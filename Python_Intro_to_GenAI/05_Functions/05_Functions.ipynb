## Returning multiple values

```python
def calculate_statistics(numbers):
    total = sum(numbers)
    avg = total / len(numbers)
    minimum = min(numbers)
    maximum = max(numbers)

    return total, avg, minimum, maximum
```

```python
marks = [70, 80, 90, 60, 85]
```

```python
total, avg, minimum, maximum = calculate_statistics(marks)
```

```python
print(total)
```

```python
print(avg)
```

```python
print(minimum)
```

```python
print(maximum)
```

## Keyword argument and positional arguments

```python
def describe_student(name, age, course):
    print(f"Name : {name}")
    print(f"Age : {age}")
    print(f"course : {course}")
```

```python
describe_student("Desmond", 30, "Data Science")
```

```python
describe_student("Software Engineering", "Annex", 30)
```

```python
# Key word argument
describe_student(
    course = "Software Engineering",
    name  = "Annex",
    age = 35
)
```

```python
# Example of a data science
def clean_sales_data(sales):
    cleaned_sales = []

    for amount in sales:
        if amount is not None and amount >= 0:
            cleaned_sales.append(amount)

    return cleaned_sales
```

```python
sales = [1200, 2500, None, -300, 1900, 0, 900]
```

```python
cleaned = clean_sales_data(sales)
```

```python
print("Original Data: ", sales)
print("Cleaned Data: ", cleaned)
```

## 2. Arguments and Return Values


```python
def greet_person(name):
    print(f"Hello, {name}!")

greet_person("Alice")
greet_person("Bob")

def add(a, b):
    return a + b

result = add(5, 3)
print("5 + 3 =", result)

# Multiple return values (returns a tuple)
def min_max(data):
    return min(data), max(data)

lo, hi = min_max([4, 2, 9, 1, 7])
print(f"Min: {lo}, Max: {hi}")
```

## 3. Default and Keyword Arguments


```python
def introduce(name, language="Python"):
    print(f"Hi, I'm {name} and I code in {language}.")

introduce("Alice")               # uses default
introduce("Bob", "JavaScript")   # overrides default

# Keyword arguments — order does not matter
def describe(name, age, city):
    print(f"{name}, {age}, from {city}")

describe(age=25, city="Accra", name="Alice")
```

## 4. Lambda Functions (Anonymous Functions)

A `lambda` is a short one-line function without a name.


```python
square = lambda x: x ** 2
print(square(5))    # 25

multiply = lambda a, b: a * b
print(multiply(4, 7))  # 28

# Lambdas with sorted(), map(), filter()
students = [("Alice", 85), ("Bob", 72), ("Charlie", 91)]
by_score = sorted(students, key=lambda s: s[1], reverse=True)
print("By score:", by_score)

nums = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x*2, nums))
evens   = list(filter(lambda x: x%2==0, nums))
print("Doubled:", doubled)
print("Evens:  ", evens)
```

## 5. Scope (Local vs Global)


```python
message = "I am global"

def show():
    local_var = "I am local"
    print(message)   # can read global
    print(local_var) # only visible inside

show()
# print(local_var)  ← NameError if uncommented

# Use global to MODIFY a global variable
counter = 0
def increment():
    global counter
    counter += 1

increment(); increment(); increment()
print("Counter:", counter)
```

## 7. Reusable Functions — Real Example


```python
def calculate_bmi(weight_kg, height_m):
    """Calculate BMI given weight in kg and height in metres."""
    return round(weight_kg / (height_m ** 2), 1)

def classify_bmi(bmi):
    if bmi < 18.5: return "Underweight"
    elif bmi < 25: return "Normal weight"
    elif bmi < 30: return "Overweight"
    else:          return "Obese"

bmi = calculate_bmi(70, 1.75)
print(f"BMI: {bmi} — {classify_bmi(bmi)}")
```
