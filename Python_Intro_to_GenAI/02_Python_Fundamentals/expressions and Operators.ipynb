# Identity Operators 
- These test if 2 variables refer to the same object in memory
`is` and `is not`

```python
a = [1, 2, 3]
b = a
```

```python
print(a is b) # True
```

```python
x = [1, 2, 3]
```

```python
print(x is a)
```

```python
print(x is b)
```

## == vs is

```python
a = [1, 2, 3]
b = [1, 2, 3]
```

```python
print(a == b) # True
```

```python
print(a is b) # False
```

```python
# is not
a = [1, 2]
b = [1, 2]

print(a is not b) # True
```

```python
value = None
if value is None:
    print("No Value")
```

# Membership Operators
- They determine whether something exists in a sequence or a collection

    - in 
    - not in

```python
fruits = ["apple", "banana", "Orange"]

print("apple" in fruits)
```

```python
"mango" in fruits
```

```python
"mango" not in fruits
```

```python
"Banana" in fruits
```

## Membership with strings

```python
text = "Python Data Science"

print("Python" in text)
```

```python
"Java" in text
```

```python
"Java" not in text
```

## Membership with Dictionaries

```python
student = {
    "name" : "Desmond",
    "age" : 30,
    "course" : "Data Science"
}
```

```python
# Check the key name in the dictionary
"name" in student
```

```python
"Desmond" in student
```

```python
# Checking for values
"Desmond" in student.values() # True
```
